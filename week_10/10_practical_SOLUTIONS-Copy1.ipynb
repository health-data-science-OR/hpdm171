````python
class Patient:
    def __init__(self, patient_id, age, seed=42):
        self.patient_id = patient_id
        self.age = age
        self.rng = np.random.default_rng(seed)

    def generate_length_of_stay(self, mean_los):
        """Return a simulated length of stay in days."""

        # Adjust the mean for patients older than 65
        if self.age > 65:
            mean_los = mean_los * 2

        # Sample and return a length of stay
        return self.rng.exponential(scale=mean_los)
````

````python
# Create two patients with different ages
younger_patient = Patient(patient_id=1, age=30, seed=42)
older_patient = Patient(patient_id=2, age=75, seed=101)

# Generate a length of stay for each patient
younger_los = younger_patient.generate_length_of_stay(mean_los=5)
older_los = older_patient.generate_length_of_stay(mean_los=5)

# Print the results
print(f"Younger patient's simulated stay: {younger_los:.2f} days")
print(f"Older patient's simulated stay: {older_los:.2f} days")
````

````python
class Hospital:
    '''
    Encapsulates a hosptial made up of wards...
    '''
    def __init__(self, n_patients, age_ranges):
        self.wards = [new_ward(i, n_patients[i], age_ranges[i]) 
                               for i in range(len(n_patients))]
    
    def add_patient(self, ward_id, patient):
        self.wards[ward_id].add_patient(patient)
        
    def add_ward(self, n_patients=0, age_range=(60, 95)):
        to_add = new_ward(len(self.wards), n_patients, age_range)
        self.wards.append(to_add)
    
    @property
    def n_patients(self):
        return sum([ward.n_patients for ward in self.wards])
    
    @property
    def n_wards(self):
        return len(self.wards)
````
