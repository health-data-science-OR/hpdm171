## Guided example 1: Patient age classification

> Before we proceed to exercises there are two guided examples with full solutions as a re-cap.  Read through the code and try different parameters to the functions to test your understanding.

A patient aged 18 or over is classified as an adult.
Otherwise they are classified as a child.


````python
def classify_patient_age(patient_age):
    """
    Return a patient age category.

    Parameters
    ----------
    patient_age : int

    Returns
    -------
    str
    """
    if patient_age >= 18:
        return "adult"
    else:
        return "child"
````

````python
print(classify_patient_age(5))
print(classify_patient_age(22))
````

## Guided example 2: Resting heart rate

Write a function that returns `True` if a resting heart rate is within the normal range 60 to 100 bpm inclusive.


````python
def heart_rate_in_normal_range(heart_rate_bpm):
    """
    Return True if resting heart rate is within the normal range.

    Parameters
    ----------
    heart_rate_bpm : int or float

    Returns
    -------
    bool
    """
    return heart_rate_bpm >= 60 and heart_rate_bpm <= 100
````

````python
print(heart_rate_in_normal_range(72))
print(heart_rate_in_normal_range(110))
````
