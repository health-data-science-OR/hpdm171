## 1. Boolean Operators 

* We have seen that Python has a type `bool` that can take two values `True` or `False`
* Boolean operators return `True` and `False` (or yes/no) answers


| Operation       | Description       | Operation       | Description |
| :------------- | :----------: | -----------: | -----------:|
| `a == b`    | `a` equal to `b`              | `a != b`    | `a` not equal to `b`             |
| `a < b`     | `a` less than `b`             | `a > b`     | `a` greater than `b`             |
| `a <= b`    | `a` less than or equal to `b` | `a >= b`    | `a` greater than or equal to `b` |


* **Tip**: Remember that the `=` operator is for **assignment** 
* use `==` when you want to **compare** equality

````python
age = 25
print(age < 40)
print(age >= 40)
print(age == 25)
````

## 2. Conditional Logic

* Most code makes extensive use of **if then** statements
* This makes use of the `if`, `elif`, and `else` keywords   

```python
if <Boolean Operation 1>:
    do something
elif <Boolean Operation 2>:
    do something different
else:
    take default action
``` 

### 2.1 A simple example of an `if` then `else` statement
* Is the patient classed as an adult or a child?

````python
patient_age = 5

if patient_age >= 18:
    print("Patient is classified as an adult")
else:
    print("Patient is classified as a child")
````

````python
patient_age = 5

if patient_age >= 18:
print("Patient is classified as an adult")
else:
    print("Patient is classified as a child")
````

### 2.2 A more complex **if then**

````python
# Define safe resting heart rate range (beats per minute)
lower_limit = 60
upper_limit = 100

# Measured resting heart rate for a patient
heart_rate = 110

if heart_rate >= lower_limit and heart_rate <= upper_limit:
    print('Heart rate is within the normal resting range')
else:
    print('Heart rate is outside the normal resting range')
````

### 2.3 Else if = `elif`

We may have multiple if statements chained together. 

````python
bmi = 32.0

# Python checks each condition in order and stops at the first one that is True
if bmi < 0:
    print('Error. Please enter a BMI greater than zero')
elif bmi < 18.5:
    print('BMI category: Underweight')
elif bmi < 25:
    print('BMI category: Normal weight')
elif bmi < 30:
    print('BMI category: Overweight')
else:
    # If none of the above conditions are True, BMI must be >= 30
    print('BMI category: Obese')
````

## 3. Functions and if statements

* We could also use a function within an `if` statement

````python
def heart_rate_in_normal_range(heart_rate_bpm):
    """
    Return True or False to indicate if
    resting heart rate is within the normal range.

    Parameters
    ----------
    heart_rate_bpm : int or float
        Resting heart rate in beats per minute.

    Returns
    -------
    bool
    """
    # Normal resting heart rate range: 60–100 bpm
    return heart_rate_bpm >= 60 and heart_rate_bpm <= 100
````

````python
heart_rate = 110

if heart_rate_in_normal_range(heart_rate):
    print('Heart rate is within the normal resting range')
else:
    print('Heart rate is outside the normal resting range')
````

* We can also use the `not` operator in Python

````python
heart_rate = 110

if not heart_rate_in_normal_range(heart_rate):
    # this branch is now executed if function returns False
    print('Heart rate is outside the normal resting range')
else:
    print('Heart rate is within the normal resting range')
````

## 4. Nested if statements

* To help with complex logic you can nest if-statements inside each other.

````python
def ed_reattendance_risk_score(days_since_last_attend, n_attends_last_year):
    """
    Toy example: return a simple numeric risk score for ED re-attendance.

    Higher score if:
    - The patient attended very recently (within 7 days), and
    - They are a frequent attender (>= 5 visits in the last year).

    Parameters
    ----------
    days_since_last_attend : int
        Days since the patient's last ED attendance.
        Use a large number (e.g. 999) if no prior attendance on record.
    n_attends_last_year : int
        Number of ED attendances in the last 12 months.

    Returns
    -------
    float
        Re-attendance risk points (toy score for teaching).
    """
    if days_since_last_attend <= 7:
        # Recently attended – now check how frequent an attender they are
        if n_attends_last_year >= 5:
            return 2.0  # recent + frequent
        else:
            return 1.0  # recent but not a frequent attender
    else:
        # Not a recent attendance – separate rule based on frequency alone
        if n_attends_last_year >= 5:
            return 1.0  # frequent but not recent
        else:
            return 0.0  # neither recent nor frequent
````

````python
risk_score = ed_reattendance_risk_score(
    days_since_last_attend=3,
    n_attends_last_year=6
)

print(f"Patients risk score is {risk_score}")
````

## 5. An alternative to nested conditionals

* We could create boolean flags first and then have a simpler `if then` structure.

````python
def ed_reattendance_risk_score(days_since_last_attend, n_attends_last_year):
    """
    Toy example: return a simple numeric risk score for ED re-attendance
    based on recency and frequency of past visits.
    """
    # Use boolean expressions to store True/False "flags"
    is_recent = days_since_last_attend <= 7
    is_frequent = n_attends_last_year >= 5

    # Then look up the risk score based on the combination of flags
    if is_recent and is_frequent:
        return 2.0          # recent + frequent
    elif is_recent and not is_frequent:
        return 1.0          # recent only
    elif (not is_recent) and is_frequent:
        return 1.0          # frequent only
    else:
        return 0.0          # neither recent nor frequent
````

````python
risk_score = ed_reattendance_risk_score(
    days_since_last_attend=3,
    n_attends_last_year=6
)

print(f"Patients risk score is {risk_score}")
````

## 6. Introduction to iterating over data using **for loops**

**Algorithms often need to do the same thing again and again.**

For example, a healthcare assistant may need to record the same observation for several patients.

**Recording observations algorithm**

1. Go to patient 1
2. Record blood pressure
3. Save the reading
4. Go to patient 2
5. Record blood pressure
6. Save the reading
7. Go to patient 3
8. Record blood pressure
9. Save the reading


**A better way to do this in code is to use a LOOP**

Do the following 3 steps, 3 times:

1. Go to the next patient
2. Record blood pressure
3. Save the reading



* There are two types of loop in Python
* `for` loops and `while` loops
* We generally use `while` if we do not know the number of iterations in advance
* We generally use `for` if we know the number of iterations in advance

### 6.1 A basic `for` loop

A `for` loop is useful when **we know the number of iterations in advance**.  Often we use `for` in combination with `range`

> Note that we use `range(3)` and the counter `patient_num` starts from 0 and ends at 2.

````python
for patient_num in range(3):
    print(f"Go to Patient {patient_num}")
    print(f"Record Blood Pressure for Patient {patient_num}")
    print(f"Save reading for Patient {patient_num}")
````

If we wanted to label patients 1 to 3 then we would need to use `range(1, 4)`. So we read this as create a range object from 1 up to and excluding 4. 
`
> It is a common mistake to forget that in end means **up to and excluding**  in `range(start, end)`.

````python
for patient_num in range(1, 4):
    print(f"Go to Patient {patient_num}")
    print(f"Record Blood Pressure for Patient {patient_num}")
    print(f"Save reading for Patient {patient_num}")
````

### 6.2 Watch out for python whitespace rules!

### 6.3 Calling a function from within a `for` loop

````python
def ward_round_task(patient_number):
    """
    For multiples of three print "blood pressure check"
    and for the multiples of five print "medication review". 
    For numbers which are multiples of both three 
    and five print "blood pressure check + medication review".
    otherwise print 'routine review'

    Note this is the FizzBuzz function - look it up!
    
    Parameters:
    ----------
    n: patient_number
        the patient number to test
    """
    if patient_number % 3 == 0 and patient_number % 5 == 0:
        print("blood pressure check + medication review")
    elif patient_number % 3 == 0:
        print("blood pressure check")
    elif patient_number % 5 == 0:
        print("medication review")
    else:
        print("routine review")
````

````python
limit = 15

for patient_number in range(1, limit + 1):
    ward_round_task(patient_number)
````

### 6.4 Looping over functions that return a value
Rather than printing out a value a function is more likely to return a value to you within a loop.  For example we might store the results of the function call in a list and use that at a later stage in our code.

````python
def ward_round_task(patient_number):
    """
    Modified function that returns str rather than
    printing it out.
    
    Parameters:
    ----------
    n: patient_number
        the patient number to test

    Returns:
    -------
    str
    """
    if patient_number % 3 == 0 and patient_number % 5 == 0:
        return "blood pressure check + medication review"
    elif patient_number % 3 == 0:
        return "blood pressure check"
    elif patient_number % 5 == 0:
        return "medication review"
    else:
        return "routine review"
````

````python
limit = 5
results = [] # empty list that we will append to

for patient_i in range(1, limit + 1):
    patient_result_i = ward_round_task(patient_i)
    results.append(patient_result_i)

# we have stored the results 
print(results)
print(results[0])
````

## 7. While Loops

* All while loops have the same structure
* You use the `while` keyword 
* Followed by a boolean operation (e.g. `age <= 18`)
* Beware of infinite loops!
* Let's test a more complex while loop using a function

* A `while` loop is useful when we do **not** know in advance how many times a process must repeat.

* Let's take an example where we have a patient and every hour we monitor that patient's oxygen saturation until it reaches a safe target.

````python
spo2 = 88  
hour = 0
target_spo2 = 94
````

````python
# A while loop does a conditional test like an if-statement
while spo2 < target_spo2:
    print(f"Hour {hour}: oxygen saturation is {spo2}%")

    # note that the thing we are testing is updated inside the loop
    spo2 += 2
    hour += 1

print(f"Target reached at hour {hour}: oxygen saturation is {spo2}%")
````

## 8. Nested Loops

* `for` and `while` loops can be nested within each other.
* Think of nested loops as a 'loop of loops'
* Remember that for each iteration outer loop will consist of multiple iterations of an inner loop
* **Don't panic** if you do not understand straight away!

### 8.1. A basic example with an outer and inner loop

````python
for outer_index in range(3):
    print(f'Outer loop iteration: {outer_index}')
    
    for inner_index in range(5):
        print(f'\tInner loop iteration: {inner_index}')
````

### 8.2 An inner loop that relates to the outer

* Let's use a nested `for` loop to create the pattern below


```python
1
12
123
1234
12345
```

To do this we need to use the outer loop counter as part of the inner loop:

````python
for outer_index in range(1, 6):
    # remember this is a loop of loops. 
    # the loop below execute all iterations each 
    # time the outer loop iterates
    for inner_index in range(1, outer_index + 1):
        
        # we use the end='' option of print so 
        # that it prints on the same line as previous
        print(inner_index, end='')
    
    # just for a new line
    print('')
````

## 9. Iterating over Python lists

We can also iterate over a `list`.  Typically this is with a `for` loop.  You can do this directly and do not need to use the `range` object.

````python
results = [10.0, 2.5, 25.1, 35.0]

for result_i in results:
   print(result_i) 
````

Sometimes when you loop through a list, you need both the **value** from the list *and* the **index** (the position or number) of that item. 

While you could manage a counter variable manually, Python provides a built-in function called `enumerate()`. It takes a list and returns pairs of `(index, value)`. By default, the index starts at `0`, but you can tell it to start at `1` if you are counting real-world items like patients.

Here is an example printing patient names alongside a patient ID number:

````python
# Create a list of patients
patients = ["Phil", "Sarah", "Becky", "Reggie"]

# Loop through the list, starting the count at 1
for patient_id, name in enumerate(patients, start=1):
    print(f"Patient {patient_id}: {name}")
````

It is possible to iterate over multiple lists at the same time using `zip`.

````python
patients = ['Phil', 'Sarah', 'Becky', 'Reggie']
results = [10.0, 2.5, 25.1, 35.0]

# notice that when we use `zip` we have a tuple: patient_name, result
for patient_name, result_i in zip(patients, results):
   print(f"{patient_name}: {result_i}") 
````

> Watch out for silent errors when using zip. Let's assume there are 5 patients, but only 4 results in the lists:

````python
patients = ['Phil', 'Sarah', 'Becky', 'Reggie', 'Tom']
results = [10.0, 2.5, 25.1, 35.0]

# notice Tom is not included!
for patient_name, result_i in zip(patients, results):
   print(f"{patient_name}: {result_i}") 
````

````python
patients = ['Phil', 'Sarah', 'Becky', 'Reggie', 'Tom']
results = [10.0, 2.5, 25.1, 35.0]

if len(patients) != len(results):
    print("Error, lists are different lengths")
else:
    for patient_name, result_i in zip(patients, results):
        print(f"{patient_name}: {result_i}") 
````
