In this exercise, you will use `numpy` to generate a sample of values from a standard normal distribution.  You will then use the statistical functions built into `numpy` to analyse the distribution of the variable.

* Call your function and print out the results.

**Hints**:

````python
# your code here...
````

````python
# example solution

SAMPLE_SIZE = 10_000

# create a random number generator
rng = np.random.default_rng(42)

# generate numpy array of size SAMPLE_SIZE using standard normal
samples = rng.normal(size=SAMPLE_SIZE)
print(type(samples))
print(samples.shape)
````

````python
def basic_descriptives(data):
    """
    Returns mean, stdev, and 1st and 99th percentile of a 1D numpy.ndarray
    
    Assumes `data` is numpy array of floats.
    
    Parameters:
    ------------
    data: numpy.ndarray 
        numeric data to analyse
        
    Returns:
    --------
    (float, float, float, float)
    """
    mean = data.mean()
    std = data.std(ddof=1)
    per_1st = np.percentile(data, 1) 
    per_99th = np.percentile(data, 99)
    
    return mean, std, per_1st, per_99th
````

````python
mean, std, p1, p99 = basic_descriptives(samples)

print(f"Mean: {mean:.3f}")
print(f"Standard deviation: {std:.3f}")
print(f"1st percentile: {p1:.3f}")
print(f"99th percentile: {p99:.3f}")
````

**Example solution:**

It is very simple to work with `numpy` arrays containing numeric data. For example if we wanted to find all of our samples that are greater than or equal to +1.96 we use:
    

````python
# result is an array of Boolean values
result = samples >= 1.96

print(result.shape)
print(type(result))
print(result)
````

The code returns a new numpy.ndarray that contains boolean (True/False) values. The value at array index i is True if the corresponding value at index i in `samples` is >= 1.96 and False otherwise. If we had used a python List we would have needed to loop through all of the list items and perform the check ourselves.

Let's create some generalised functions to return the probabilities that a value is greater or less than a user specified value in our data set.

To do that we need to know that

```python
False == 0
True == 1
```

Therefore we can take the sum of our boolean array to find out how many array elements are greater or less than a user specified value. i.e.

````python
(samples >= 1.96).sum()
````

````python
def prob_greater_than_or_equal_to(data, x):
    '''
    Return the proportion of the dataset that is greater than or equal to x
    
    Parameters:
    -----------
    data: numpy.ndarray 
        numeric data to analyse
    x: float
        Lower cut-off. 
        
    Returns:
    --------
    float
    '''
    # alternative: return (data >= x).mean()
    return (data >= x).sum()/data.shape[0]


def prob_less_than_or_equal_to(data, x):
    '''
    Return the proportion of the dataset that is less than or equal to x
    
    Parameters:
    -----------
    data: numpy.ndarray 
        numeric data to analyse
    x: float
        Upper cut-off. 
        
    Returns:
    --------
    float
    '''
    # alternative: return (data <= x).mean()
    return (data <= x).sum()/data.shape[0]

p1 = prob_greater_than_or_equal_to(samples, 1.96)
p2 = prob_less_than_or_equal_to(samples, -1.96)

print(p1, p2)
print(1 - (p1 + p2))
````

Our test of these functions shows us that around 95% of data lie between points -1.96 and +1.96 (which again we would expect with the standard normal).

## Exercise 7: Simulate total bed-days without Python loops

> In this exercise you are given some python code that uses two `for` loops that operate on a `numpy` array. Just like in the bootstrap example your task is to eliminate those loops and use `numpy` vectorized operations to improve efficiency.

A hospital wants to explore the total bed-days associated with 100 respiratory admissions.

In this exercise we will assume that each patient's length of stay is an independent draw from an exponential distribution with a mean of 5 days. 

The supplied function `bed_days_python` simulates a length of stay for each patient and adds the stays together. It repeats this experiment many times, returning one total per simulation.

- The outer loop repeats the simulation.
- The inner loop simulates individual patients.

````python
def bed_days_python(n_simulations, n_patients, seed=42):
    """Return simulated total bed-days using two Python loops."""
    rng = np.random.default_rng(seed)
    totals = []

    # outer loop: run multiple simulations
    for simulation in range(n_simulations):
        total = 0

        # inner loop: simulate multiple patients LoS
        for patient in range(n_patients):
            length_of_stay = rng.exponential(scale=5)
            total += length_of_stay

        totals.append(total)

    return np.array(totals)
````

````python
n_simulations = 1_000
n_patients = 100

python_totals = bed_days_python(n_simulations, n_patients)
print("Result shape:", python_totals.shape)
print("First five totals:", python_totals[:5])
````

### Example solution: remove the inner loop

Instead of drawing one patient's stay at a time, generate a vector containing all patient stays for one simulation and sum the vector.

````python
def bed_days_inner_numpy(n_simulations, n_patients, seed=42):
    """Remove the patient loop, retaining the simulation loop."""
    rng = np.random.default_rng(seed)
    totals = np.empty(n_simulations)

    for simulation in range(n_simulations):
        lengths_of_stay = rng.exponential(scale=5, size=n_patients)
        totals[simulation] = lengths_of_stay.sum()

    return totals
````

### Example solution: remove both loops

Generate all lengths of stay in one call. The resulting matrix has shape `(n_simulations, n_patients)`. Each row is one simulated group of admissions. Summing across the columns (`axis=1`) gives one total per row.

````python
def bed_days_numpy(n_simulations, n_patients, seed=42):
    """Return simulated total bed-days without Python loops."""
    rng = np.random.default_rng(seed)
    lengths_of_stay = rng.exponential(
        scale=5,
        size=(n_simulations, n_patients),
    )
    return lengths_of_stay.sum(axis=1)
````

````python
python_totals = bed_days_python(n_simulations, n_patients, seed=42)
inner_totals = bed_days_inner_numpy(n_simulations, n_patients, seed=42)
numpy_totals = bed_days_numpy(n_simulations, n_patients, seed=42)

np.testing.assert_allclose(python_totals, inner_totals)
np.testing.assert_allclose(python_totals, numpy_totals)
assert numpy_totals.shape == (n_simulations,)
print("All three implementations produce matching results.")
````
